```python
# 1. Instale a biblioteca (se ainda não o fez na sessão)
!pip install gaitmap -q

import numpy as np
import pandas as pd
# Importe o algoritmo que EXISTE na sua versão
from gaitmap.stride_segmentation import RoiStrideSegmentation
from gaitmap.utils.array_handling import to_stride_list

# 2. Simule ou carregue seus dados
sampling_rate_hz = 100
n_samples = 2000
time = np.arange(n_samples) / sampling_rate_hz
gyr_s = np.sin(2 * np.pi * 2 * time) * 10 + np.random.rand(n_samples) * 0.5
data = pd.DataFrame({"gyr_s": gyr_s}, index=time)

# 3. Crie uma "Região de Interesse" (ROI)
# Para este exemplo, vamos dizer que a caminhada acontece no trecho inteiro.
# A lista de ROIs tem o formato [[inicio_1, fim_1], [inicio_2, fim_2], ...]
# Onde os valores são em amostras.
regions_of_interest = [[0, len(data)]]

# 4. Inicialize e execute o algoritmo
# Ele precisa dos dados e da lista de ROIs
segmenter = RoiStrideSegmentation()
segmenter = segmenter.segment(data, regions_of_interest, sampling_rate_hz=sampling_rate_hz)

# 5. Inspecione os resultados
print("Número de passadas detectadas:", len(segmenter.stride_list_))
print("\nAs 5 primeiras passadas detectadas:")
print(segmenter.stride_list_.head())
```

```python

# 1. Simular dados de um sensor IMU (giroscópio)
# Em um caso real, você carregaria seus próprios dados de um arquivo CSV ou outro formato.
# Aqui, criamos um sinal senoidal simples para simular 10 passadas.
sampling_rate_hz = 100  # O sensor grava 100 amostras por segundo
n_samples = 720
time = np.arange(n_samples) / sampling_rate_hz
# Simulando o sinal do giroscópio no eixo sagital (para frente/trás)
gyr_s = np.sin(2 * np.pi * 2 * time) * 10 + np.random.rand(n_samples) * 0.5
data = pd.DataFrame({"gyr_s": gyr_s}, index=time)

# 2. Inicializar o algoritmo de segmentação
# GsdIluz é uma boa escolha para começar.
gsd = GsdIluz()

# 3. Executar a segmentação nos dados
# O método 'segment' recebe os dados e a taxa de amostragem.
gsd = gsd.segment(data, sampling_rate_hz=sampling_rate_hz)

# 4. Inspecionar os resultados
# O resultado é uma lista de passadas (stride_list).
# Cada entrada no dicionário representa uma passada, com seu 'start' e 'end' em amostras.
print("Número de passadas detectadas:", len(gsd.stride_list_))
print("\nAs 5 primeiras passadas detectadas (início, fim e id):")
# O resultado é um DataFrame do pandas
print(gsd.stride_list_.head())
```
